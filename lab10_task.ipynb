## Step 1: Import Libraries and Load Data

```python
# Import required libraries
import pandas as pd
import numpy as np
```

```python
# Load the dataset (continuing from Lab 9)
df = pd.read_csv('car_sales_data.csv')
print("Dataset loaded successfully!")
print(f"Total records: {len(df)}")
print(f"Shape: {df.shape}")
```

```python
# Display initial data info
print("Initial Dataset Information:")
df.info()
```

## Step 2: Data Pre-Processing - Check Data Quality

```python
# Check first few rows
print("First 5 rows:")
df.head()
```

```python
# Check for missing values
print("Missing Values Count:")
print(df.isnull().sum())
print(f"\nTotal Missing Values: {df.isnull().sum().sum()}")
print(f"\nMissing Values Percentage:")
print((df.isnull().sum() / len(df) * 100).round(2))
```

```python
# Check data types before preprocessing
print("Data Types Before Preprocessing:")
print(df.dtypes)
```

## Step 3: Dealing with Null Values

```python
# Identify categorical and numerical columns
categorical_cols = df.select_dtypes(include=['object']).columns.tolist()
numerical_cols = df.select_dtypes(include=['int64', 'float64']).columns.tolist()

print(f"Categorical Columns: {categorical_cols}")
print(f"Numerical Columns: {numerical_cols}")
```

```python
# Fill null values in categorical columns with mode
print("Filling null values in categorical columns with mode...")
for col in categorical_cols:
    if df[col].isnull().sum() > 0:
        mode_value = df[col].mode()[0]
        df[col].fillna(mode_value, inplace=True)
        print(f"  {col}: Filled with '{mode_value}'")
```

```python
# Fill null values in numerical columns with median
print("Filling null values in numerical columns with median...")
for col in numerical_cols:
    if df[col].isnull().sum() > 0:
        median_value = df[col].median()
        df[col].fillna(median_value, inplace=True)
        print(f"  {col}: Filled with {median_value}")
```

```python
# Verify null values are handled
print("Missing Values After Handling:")
print(df.isnull().sum())
print(f"\nTotal Missing Values: {df.isnull().sum().sum()}")
```

## Step 4: Changing Datatypes of Columns to Int

```python
# Display current data types
print("Current Data Types:")
print(df.dtypes)
```

```python
# Convert float columns to int
print("Converting appropriate columns to integer type...\n")

# Check which numerical columns can be converted to int
for col in numerical_cols:
    if df[col].dtype == 'float64':
        # Check if all values are whole numbers
        if (df[col] == df[col].astype(int)).all():
            df[col] = df[col].astype(int)
            print(f"  {col}: Converted to int")
        else:
            print(f"  {col}: Kept as float (contains decimal values)")
```

```python
# Specific columns that should be integers
# Convert Engine size, Year of manufacture, and Mileage to int
columns_to_int = ['Engine size', 'Year of manufacture', 'Mileage']

for col in columns_to_int:
    if col in df.columns:
        df[col] = df[col].astype(int)
        print(f"Converted {col} to int")
```

```python
# Display data types after conversion
print("Data Types After Conversion:")
print(df.dtypes)
```
